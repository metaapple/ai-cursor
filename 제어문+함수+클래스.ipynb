```python
def max_value(a, b):
  result = 0
  if a > b:
    result = a;
  elif a < b:
    result = b;
  return result
```

```python
print(max_value(100, 200))
```

```python
def count_even(numbers): #[1, 2, 3, 4, 5]
  print(type(numbers))
  count = 0
  for x in numbers:
    if x % 2 == 0:
      count = count + 1
  return count
```

```python
def check_type(value):
    if isinstance(value, bool):
        return "boolean"
    elif isinstance(value, int):
        return "정수"
    elif isinstance(value, float):
        return "실수"
    elif isinstance(value, str):
        return "문자열"
    elif isinstance(value, list):
        return "list"
    else:
        return "기타"


print(check_type(10))          # 정수
print(check_type(3.14))        # 실수
print(check_type("Python"))    # 문자열
print(check_type(True))        # boolean
print(check_type([1, 2, 3]))   # list
```

```python
target = random.randint(1, 100)
count = 0
while True:
  num = int(input("숫자입력>> "))
  count = count + 1
  print(num)
  if num == target :
    print("정답입니다.", count)
    break; #while문 중단!!
  elif num > target:
    print("너무 큽니다.")
  elif num < target:
    print("너무 작습니다.")
```
