```python
for x in range(5):
  if x == 3:
    # break #for문 stop!
    continue #for문 계속, skip
    # pass
  if x >= 1:
    print(x + 3)
```

```python
start = 0;
while True:
  print("반복중")
  start = start + 1;
  if start > 3:
    break; #무한 반복문에서는 반드시 break되는 지점을 넣어주어야함.
```

```python
def greet(name):
  #name변수는 greet()함수 안에서 만들었으로
  #함수밖에서 쓸 수 없음.
  #함수 범위(지역)안에서만 사용 가능
  #지역변수(local)
  print(name, '안녕!!!!')
```

```python
greet('홍길동')
greet('이순신')
```

```python
def greet2(name, message="안녕하세요."):
  print(name, message)
```

```python
greet2("민수")
```

```python
greet2("민수", "반값습니다.")
```

```python
greet2(name="홍길동", message="잘가")
greet2(message="잘가", name="홍길동")
```

```python
def add(a,b):
  # print(a + b)
  # 함수처리 결과를 다음 아래줄에서
  # 계산할 때 사용하고자하는 경우
  # 그 결과값을 반환해야함.
  result = a + b;
  return result;
```

```python
print(add(10, 20)) #\return하면 함수 호출한 줄로 처리결과를 반환받음.
```

```python
answer = add(100, 200)
answer
```
